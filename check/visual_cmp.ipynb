## TT_wide

```python
TT_wide_great_times = lambda x: TT_wide(
    si, li, pzi, sf, lf, x, kp, kp * rho_H, t_in, t_out
)

TT_wide_numeric = lambda x: TT_wide_Texact(
    si, li, pzi, sf, lf, x, kp, kp * rho_H, t_in, t_out
)
```

```python
import numpy as np

import matplotlib.pyplot as plt

def plot_TT_wide(xmin, xmax, num_points=100):
    x_vals = np.linspace(xmin, xmax, num_points)
    y_great_times = [TT_wide_great_times(x) for x in x_vals]
    y_numeric = [TT_wide_numeric(x) for x in x_vals]

    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    plt.plot(x_vals, [y.real for y in y_great_times], label='TT_wide_great_times (Real)')
    plt.plot(x_vals, [y.real for y in y_numeric], label='TT_wide_numeric (Real)', linestyle='--')
    plt.xlabel('x')
    plt.ylabel('Real Part')
    plt.title('Real Part of TT_wide Functions')
    plt.legend()
    plt.grid(True)

    plt.subplot(1, 2, 2)
    plt.plot(x_vals, [y.imag for y in y_great_times], label='TT_wide_great_times (Imag)')
    plt.plot(x_vals, [y.imag for y in y_numeric], label='TT_wide_numeric (Imag)', linestyle='--')
    plt.xlabel('x')
    plt.ylabel('Imaginary Part')
    plt.title('Imaginary Part of TT_wide Functions')
    plt.legend()
    plt.grid(True)

# Example usage:
plot_TT_wide(xmin=0, xmax=10)
```
